# Project 5: Activation Functions Using Python

## Objective
Implement commonly used activation functions and observe how they transform input values.

### Tasks
1. Implement **Sigmoid, ReLU, and Tanh** activation functions.
2. Generate outputs for different input values.
3. Visualize and compare the behavior of the three activation functions.

## Step 1: Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2: Implement Activation Functions

```python
# Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# ReLU Activation Function
def relu(x):
    return np.maximum(0, x)

# Tanh Activation Function
def tanh(x):
    return np.tanh(x)
```

## Step 3: Generate Outputs for Different Input Values

```python
# Input values
inputs = np.array([-5, -3, -1, 0, 1, 3, 5])

# Calculate outputs
sigmoid_output = sigmoid(inputs)
relu_output = relu(inputs)
tanh_output = tanh(inputs)

print('Input Values:', inputs)
print('Sigmoid Output:', sigmoid_output)
print('ReLU Output:', relu_output)
print('Tanh Output:', tanh_output)
```

### Output Comparison Table

```python
print(f"{'Input':>8} {'Sigmoid':>12} {'ReLU':>12} {'Tanh':>12}")
print('-' * 48)
for x in inputs:
    print(f"{x:>8.1f} {sigmoid(x):>12.4f} {relu(x):>12.4f} {tanh(x):>12.4f}")
```

## Step 4: Visualize and Compare the Activation Functions

```python
# Create a range of input values for visualization
x = np.linspace(-10, 10, 400)

# Plot all activation functions
plt.figure(figsize=(10, 6))
plt.plot(x, sigmoid(x), label='Sigmoid')
plt.plot(x, relu(x), label='ReLU')
plt.plot(x, tanh(x), label='Tanh')

plt.title('Comparison of Activation Functions')
plt.xlabel('Input Value')
plt.ylabel('Output Value')
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.grid(True)
plt.legend()
plt.show()
```

## Conclusion

- **Sigmoid** transforms values into the range **0 to 1** and produces an S-shaped curve.
- **ReLU** converts all negative values to **0** and keeps positive values unchanged.
- **Tanh** transforms values into the range **-1 to 1** and is centered around zero.

These activation functions are important in neural networks because they introduce non-linearity and help models learn complex patterns.
