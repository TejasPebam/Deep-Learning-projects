# MSE and Cross-Entropy Loss

## Objective
To calculate Mean Squared Error (MSE) and Cross-Entropy Loss, compare the results, and understand their uses.

## Task 1: Define Actual and Predicted Values

We use actual binary class labels and predicted probabilities from a simple classification problem.

```python
import numpy as np

y_true = np.array([1, 0, 1, 1, 0])
y_pred = np.array([0.9, 0.2, 0.7, 0.6, 0.1])

print("Actual Labels:", y_true)
print("Predicted Outputs:", y_pred)
```

## Task 2: Calculate MSE and Cross-Entropy Loss

```python
# Mean Squared Error
mse = np.mean((y_true - y_pred) ** 2)

# Binary Cross-Entropy Loss
eps = 1e-15
p = np.clip(y_pred, eps, 1 - eps)
cross_entropy = -np.mean(y_true*np.log(p) + (1-y_true)*np.log(1-p))

print("MSE:", mse)
print("Cross-Entropy Loss:", cross_entropy)
```

## Task 3: Compare the Results

MSE measures squared numerical error, while Cross-Entropy measures the difference between actual classes and predicted probabilities. Their numerical values should not be directly compared because they have different scales.

```python
print(f"MSE = {mse:.4f}")
print(f"Cross-Entropy = {cross_entropy:.4f}")

print("\nMSE: commonly used for regression.")
print("Cross-Entropy: commonly used for classification.")
```

## Uses

- **MSE:** Used mainly for regression, such as predicting house prices or temperature.
- **Cross-Entropy:** Used mainly for classification, such as spam detection or cat/dog classification.

## Conclusion
MSE calculates the average squared prediction error. Cross-Entropy is more suitable for classification because it strongly penalizes confident incorrect predictions. Therefore, the appropriate loss function depends on the type of machine learning problem.
